```python
from ipywidgets import interact
import pandas as pd

pd.options.plotting.backend = 'plotly'
```

```python
df = pd.read_csv("dataset.csv")
df.head(10)
```

```python
@interact
def filter_company(company = list(df.Company.unique())):

    return df[df.Company == company].head()
```

```python
@interact
def filter_rating(rating = (0, 5, 0.5)):

    new_df = df[df.Rating>rating]
    display(new_df.head())
```

```python
@interact
def plot_city(city = list(df.City.unique())):

    filtered_df = df[df.City == city]
    
    ## create the plot on filtered data
    fig = filtered_df.plot(kind = "scatter", 
                           x = "Salary", 
                           y = "Rating", 
                           color = "Status",
                           size = "Credits")
    fig.show()
```
